```python
from plot_tools.plot_fields import plot_current_v_time, plot_center_density_v_time
import numpy as np
```

```python
dir = "diags_output/"

ICP_mag = 500.0  # A/m^2, ICP current density amplitude
ICP_freq = 10e6  # Hz, ICP frequency
target_current = lambda t: ICP_mag * np.sin(2 * np.pi * ICP_freq * t)

J_total = np.load(dir + "J_total_y.npy")
times = np.load(dir + "collection_times.npy")
Ne = np.load(dir + "N_e.npy")

fig, ax = plot_current_v_time(J_total, times, target_current)
fig, ax = plot_center_density_v_time(Ne, times)
```
